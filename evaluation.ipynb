## Evaluation metrics

```python
# code pour faire l'évaluation de nos modèles
import numpy as np
```

```python
y_true, y_pred = np.random.randint(0, 2, 10), np.random.randint(0, 2, 10)
y_true, y_pred # random labels
```

```python
# accuracy
from sklearn.metrics import accuracy_score
accuracy_score(y_true, y_pred)
```

```python
# recall
from sklearn.metrics import recall_score
recall_score(y_true, y_pred)
```

```python
# precision
from sklearn.metrics import precision_score
precision_score(y_true, y_pred)
```

```python
# f1_score
from sklearn.metrics import f1_score
f1_score(y_true, y_pred)
```

```python
# average precision
from sklearn.metrics import average_precision_score
average_precision_score(y_true, y_pred)
```

```python
# ROC
from sklearn.metrics import roc_auc_score
y_probs = np.random.rand(10)
roc_auc_score(y_true, y_probs) # y_pred should be probabilities! not predicted
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import roc_curve
# Returns arrays of coordinates
fpr, tpr, thresholds = roc_curve(y_true, y_probs)

# Now you use those coordinates to plot the ROC
plt.plot(fpr, tpr)
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.show()
```
